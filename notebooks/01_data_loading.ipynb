## Data Loading

````python
def load_sales_data(filepath='../data/sample_sales.csv'):
    """Load sales data from CSV.
    
    Best Practice: Use relative paths for repo data,
    mount points for production data.
    
    Args:
        filepath: Path to data file
        
    Returns:
        DataFrame with sales data
    """
    df = pd.read_csv(filepath)
    return df
````
