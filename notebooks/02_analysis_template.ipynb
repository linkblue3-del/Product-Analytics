## 3. Analysis

Describe what this analysis does and why.

```python
# Your analysis code here

# Calculate total revenue
total_revenue = (df['price'] * df['quantity']).sum()
print(f"Total Revenue: ${total_revenue:.2f}")

# Calculate revenue by category
revenue_by_category = df.groupby('category').apply(lambda x: (x['price'] * x['quantity']).sum())
print("Revenue by Category:")
print(revenue_by_category)

# Calculate average order value
avg_order_value = (df['price'] * df['quantity']).groupby(df['order_id']).sum().mean()
print(f"Average Order Value: ${avg_order_value:.2f}")
# Best Practice: Break into logical sections with markdown
```
